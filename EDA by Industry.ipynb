NAIC code reference is in the file /'NAIC code reference.txt'. For this notebook I focused on number of establishments by industry and region.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from Biz_Set_Functions import *
df = pd.read_pickle('full_clean_df.pkl')
```

```python
df['naics'].value_counts()
```

```python
df.naics=df.naics.astype('category')
```

```python
"""
The bottom 11 industries (the absolute smallest) were dropped from the rest of this analysis to focus on the top 15.
"""

sns.lineplot(data=df,x='year',y="establishments",hue="naics",estimator='sum', errorbar=None)
```

```python
"""
Because of the huge difference in numbers between metro area and the rest of the regions, it makes more sense to plot
and study the industries by region.
"""

mountain = df[df['region'] == 'Mountain']

eastern = df[df['region'] == 'Eastern']

metro = df[df['region'] == 'Metro Area']

southern = df[df['region'] == 'Southern']

northern = df[df['region'] == 'Northern']

western = df[df['region'] == 'Western']
```

```python
plt.figure(figsize=[10,8])
sns.lineplot(data=mountain, x='year',y="establishments",style='naics', hue='naics',estimator='sum', errorbar=None, 
             palette='colorblind', markers=True)
plt.legend(title='NAICS', loc='upper left')
plt.title(label="Establishments in the Mountain Region by NAICS codes")
```

## General observations

44: Retail Trade and 62: Health Care and Social Assistance, were consistently the top 2 industries in each region. There's been a consistent downward trend in these industries across the regions as well, the mountain area saw the least drastic decline and the metro area saw the biggest dip in 2020 but quickly improved the next year(though still nowhere near their highest numbers)

In a distant third, 56: Administrative,Support, Waste Management and Remediation Services has remained relatively steady 
across the decade

Across all regions, 31: Manufacturing food and beverage is the industry that saw the biggest upward trend across the decade experiencing a big bump in numbers from 2019 to 2020 which many more estabishments in 2022 than in the beginning of the decade. It is the only industry that is doing better than in 2013.

All significant changes occured around the 2020 mark, suggesting the pandemic had a much bigger effect on business establishments across the island than Hurricane Maria.
